#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
basketball_table <- basketball_tourney |>
    summarize(conf_mean = mean(conf_wins_pct),
             reg_mean = mean(reg_wins_pct),
             conf_min = min(conf_wins_pct),
             conf_max = max(conf_wins_pct),
             reg_min = min(reg_wins_pct),
             reg_max = max(reg_wins_pct),
            seed_min = min(seed),
             seed_max = max(seed)) 


basketball_table
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
basketball_na <- basketball |>
    summarize(total_n = n(),
            conf_na_count = sum(is.na(conf_wins_pct)), #the only one with na values because every school plays in regular
             reg_na_count = sum(is.na(reg_wins_pct)),
             tourney_na_count = sum(is.na(tourney_finish)),
             school_na_count = sum(is.na(school)),
             year_na_count = sum(is.na(year)),
             seed_na_count = sum(is.na(seed))) |>
    mutate(total_dropped = total_n - nrow(basketball_tourney)) #the total of all na values dropped

basketball_na
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

- Use `ggplot` to plot the distribution of one of the numeric variables in your tidied data set.

- Use `ggplot` to create a visualization showing the relative proportions of each category/level for one of categorical variable in your tidied data set.

```r
options(repr.plot.width = 12, repr.plot.height = 10)

#so the underscores are removed, solely for plotting purposes
basketball_clean_names <- basketball_tourney |>
    mutate(tourney_finish = fct_recode(tourney_finish, "Opening Round Loss" = "opening_round_loss", 
                                       "First Round Loss" = "first_round_loss", 
                                       "Second Round Loss" = "second_round_loss",
                                       "Top 16 Loss" = "top_16_loss", 
                                       "Top 8 Loss" = "top_8_loss", 
                                       "Top 4 Loss" = "top_4_loss", 
                                       "Top 2 Loss" = "top_2_loss", 
                                       "Champ" = "champ"))

#histogram of conference wins percentage distribution of each tourney_finish
conf_wins_distribution <- basketball_clean_names |>
    ggplot(aes(x = conf_wins_pct, fill = tourney_finish)) +
    geom_histogram(bins = 30) + #to show distribution of the percentage of conference wins
    facet_grid(rows = vars(tourney_finish)) + #for each tourney_finish
    labs(x = "Conference Win Percentage %",
         y = "Number of Teams",
         fill = "Tourney Finish Result"
    ) +
    theme(text = element_text(size = 15))
conf_wins_distribution
```

```r
options(repr.plot.width = 12, repr.plot.height = 7)

#plots each tourney_finish category and shows the percentage of each category relative to total tourney_finish count
tourney_finish_plot <- basketball_clean_names |>
    ggplot(aes(x = tourney_finish, fill = tourney_finish)) +
    geom_bar(aes(y = after_stat(count / sum(count)))) + #calculates the percentage value of a category of tourney_finish relative to total count of tourney_finish
    scale_y_continuous(labels = scales::percent) + #makes the y axis show as % instead of 0 - 1
    labs(x = "Tourney Finish Result",
         y = "Proportion of Tourney Finish Results %",
         fill = "Tourney Finish Result"
    ) +
     theme(text = element_text(size = 15), axis.text.x = element_text(angle = 30, hjust = 1, vjust = 1)) #adjusts x axis text to not overlap

tourney_finish_plot
```
